```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.info()
```

```python
plt.scatter(titanic['age'], titanic['fare'])
plt.xlabel('Edad')
plt.ylabel('Tarifa')
plt.title('Relación entre edad y tarifa')
```

```python
sns.boxplot(x = 'pclass', y = 'age', data = titanic, palette = 'Set2')
plt.title('Distribución de edad por supervivencia')
plt.show()
```

```python
sns.barplot(x = 'survived', y = 'age', data =  titanic, errorbar= None, palette='muted', hue = 'pclass', legend = True)
```
